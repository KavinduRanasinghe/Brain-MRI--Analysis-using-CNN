#### CNNs for Classification

Convolutional Neural Networks are used in a wide range of classification tasks,because of their automatic learning ability from heirachical features from pictures. CNNs can be guided and trained to identify the differences between abnormal brain MRI scans and healty normal brain scans. Here we have discussed about for types of brain tumor classes.
* Gliomas Tumor
* Meningiomas Tumor
* No Tumor
* Pitiutary Tumor

If we dive in to tumor types we can identify these 4 types have difference ascpet. Therefore CNNs extract the features from the images and identify seperate tumor types individually.


##### Gliomas Tumors
Glimas type of brain tumors originated in the glial cells, which acts as the support and the protection of the neurons in the brain. It commonly locates in the cerebral hemispheres. This type are well known from their irregular and diffuse growth patterns,which can make them hard to treat.


##### Meningiomas Tumors
Meningiomas grow from the meninges, which is the layers of tissue that cover and protect the brain and spinal cord. They are often placed on the outer surface of the brain or along the spine. Meningiomas usually grow slowly and don’t conquer brain tissue, but they can press on nearby parts of the brain and cause problems.

##### Pituitary Tumors
Pituitary tumors start in the pituitary gland, a small organ at the bottom of the brain that controls hormones in the body. These tumors are often small and stay in one place, but they can cause issues like blurry vision or hormone changes. Most pituitary tumors are not cancer effective but may still need treatment.

##### No Tumor
The "No Tumor" category refers to brain scans that don’t show any abnormal signs of a tumor. These scans look normal, withiout any abnormal growths or changes in the brain MRIs. CNNs learn these normal patterns from images, helping them tell the difference between healthy brains and ones with tumors.

#### CNNs for Segmentation Using U-Net Architecture

U-Net is a powerful convolutional neural network (CNN) architecture specifically designed for brain tumor segmentation, which involves separating the tumor area from surrounding healthy brain tissue. This process is crucial for medical diagnosis and treatment planning, as it provides precise pixel-level predictions.



##### U-Net Architecture for Brain Tumor Segmentation
U-Net has a two-part structure:  
1. Encoder: This part extracts features and context from the input MRI image. It progressively reduces the image's spatial dimensions while capturing high-level details.  
2. Decoder: This part reconstructs the segmented output by progressively increasing the spatial dimensions. Skip connections between corresponding encoder and decoder layers ensure that fine details, such as tumor edges, are preserved.


##### Preprocessing
MRI images are preprocessed to ensure consistent input for the model. Steps include:  
- Resizing images to a fixed size (e.g., 256x256 pixels).  
- Normalization to scale pixel values.  
- Data augmentation using techniques like rotation, flipping, and elastic deformations to improve model robustness.  

The output is a binary mask, where tumor areas are marked differently from healthy tissue.


##### Training
U-Net is trained using pixel-level annotations, where each pixel is labeled as tumor or non-tumor. The model optimizes its parameters to minimize pixel-wise loss functions like the Binary Cross-Entropy Loss. The training process aims to accurately match the predicted segmentation mask with the ground truth mask provided by experts.


##### Advantages of U-Net for Segmentation
- High accuracy in identifying precise tumor boundaries.  
- Effective in handling limited datasets through data augmentation.  
- End-to-end learning allows simultaneous feature extraction and segmentation.


##### Disadvantages
- Requires detailed labeled segmentation masks, which are time-consuming to create.  
- Computationally intensive during training, although inference is faster.


##### Output of U-Net
The model outputs a binary mask that highlights the tumor region in the MRI scan. This pixel-wise classification provides detailed information about the tumor's location and shape, aiding in tasks like tumor volume measurement or treatment planning.



### Technique 2: Support Vector Machines (SVMs)

Support Vector Machines highly depend on feature extraction because they cannot learn about the features automatically like CNNs. In preprocessing for SVMs have to follow several main steps. it includes the feature extraction, dimensionality reduction to improve the model's performance.

Preprocessing for SVMs in Brain Tumor detection and segmentation

Before feeding the images to the SVM model, we need to be preprocessed.

We should standardize the dimensions of the images to ensure consistency. We can apply filters (like Gaussian or median filtering) to reduce noise and enhance the quality of the image. Normalize pixel intensities to ensure the range of pixel values is consistent across all images, avoiding variations due to scanning methods.

In Feature extraction ,main features such as shape , texture are mostly identified. Shape features , boundart shape are mostly important in image classification. Normalization make sure that features are on a same level scale, any single feature is not dominating the decision making process. Mostly normalized into 0 and 1 range. Dimensionality reduction is used because SVMs can be sensitive to the high dimensional feature sets. Therefore have to use dimensionality reduction techniques like PCA (Principal Component Analysis). It helps to improve the perfomance and the reduces the risk of overfitting, when dealing with complex datasets.

How SVM works Brain Tumor detection and segmentation

SVMs are supervised learning algorithms that aim to find the optimal hyperplane to separate data points into different classes. In brain tumor detection, SVMs use extracted features from MRI images to classify them based on the patterns identified. For image classification tasks, features like texture, shape, and statistical measures are fed into the SVM model. The model then classifies the image as containing a benign or malignant tumor based on these features, making it a powerful tool for diagnostic purposes. Although SVMs are not specifically designed for image segmentation, they can be applied to classify each pixel of an image. Each pixel is treated as a feature vector based on its surrounding pixel intensities, and the SVM classifies it as tumor or non-tumor.

Advantages of SVMs for Tumor Detection

* High accuracy in classification, especially when feature extraction is well-designed, making it suitable for identifying multiple tumor classes.
* Robust to overfitting, particularly with proper regularization, making it a reliable choice for small-scale datasets with multiple classes.

Disadvantages

* Requires manual feature extraction, which demands domain-specific knowledge and is time-consuming.
* Not suitable for large datasets, as SVM training time increases exponentially with the number of data points.

### Technique 3: K-Means Clustering

K Means clustering is an unsupervised ML algorithm used in medical imaging. It include the brain tumor detection and segmentation. Due to being a unsupervised learning technique, therefore K means does not want labels within the dataset. This will be a mostly helpful when we have to work with datasets without labeled data. K means can identify the patterns in data by grouping similar data points into clusters, supporting the segmentation of tumor regions in brain MRI images. The algorithm aids in identifying the presence of a tumor classification and segmenting the tumor region within MRI images.

Preprocessing for K-Means in Brain Tumor Analysis

Better preprocessing ensures accurate clustering results , critical for classification and segmentation tasks. MRIs have to resized into consistent dimensions to ensure uniformity across datasets. Noise can be removed using filters such as Gaussian or median. It enhance the overall quality of the images. In addition pixel intensity values are normalized with the range 0 to 1 . It reduce variability caused by different imaging techniques.These preprocessing steps enhance the algorithm's ability to detect meaningful patterns.

Each pixel in an MRI image is represented as a feature vector in K-Means clustering, with attributes such as,
- Intensity (brightness levels of the pixel)
- Texture (patterns in the pixel's local neighborhood)
- Color (if applicable)

Normalization helps make sure that all features are on the same scale, so that no single feature has more weight on the clustering than the others. Dimensionality reduction, like Principal Component Analysis (PCA), is used to simplify the data by reducing the number of features, making the process speedy and focusing on the most important features for clustering.

How K-Means Clustering Works

Classification Process

The algorithm groups image pixels into different clusters. These clusters could represent normal or abnormal regions. It analyzes pixel intensity and texture patterns. Abnormal tissue regions are identified from these clusters. This helps classify the image as tumor-positive or tumor-negative.

Segmentation Process

The MRI image is split into a set number of clusters (K). Each pixel is assigned to the closest cluster based on how similar it is to the cluster's center (centroid). The centroids are updated repeatedly to reduce differences within each cluster. The clusters for gray matter, white matter, cerebrospinal fluid (CSF), and abnormal tumor regions are then identified. By separating clusters with unique intensity or texture patterns, K-Means helps detect and segment tumor areas. This allows radiologists to precisely locate the tumor regions.


Advantages of K-Means in Brain Tumor Analysis:

- K-Means doesn’t require labeled data, making it useful for exploratory analysis or when there are no annotated datasets.

- It can quickly process large medical datasets, saving time.

- K-Means can both classify if a tumor is present and segment the tumor area for further study.


Disadvantages of K-Means in Brain Tumor Analysis:

- The algorithm's results can change based on the starting points, so careful tuning is needed.

- It’s hard to determine the right number of clusters (K), especially with tumors that vary in shape and size.

- K-Means assumes clusters are round, which may not work well for irregularly shaped tumor areas.

The code defines a list of tumor types and a function to rename .jpg image files in class-specific subdirectories. It checks if each file is a .jpg, then renames it using a format like class_name_counter.jpg. The counter ensures each file within a class folder gets a unique sequential name.

```python
# List of class names corresponding to the folder names
class_names = ['glioma_tumor', 'meningioma_tumor', 'no_tumor', 'pituitary_tumor']

# Function to rename images in a directory
def rename_images_in_dir(directory):
    for class_name in class_names:
        class_dir = os.path.join(directory, class_name)  # Get the path to the class folder
        if not os.path.exists(class_dir):
            continue  # Skip if the class folder doesn't exist

        # List all files in the class directory
        files = os.listdir(class_dir)

        # Counter for the images in the class
        counter = 1

        # Iterate over each file and rename it
        for file in files:
            if file.lower().endswith(".jpg"):  # Check if the file is a JPG image
                # Generate the new name based on the class name and counter
                new_name = f"{class_name}_{counter}.jpg"

                # Define full paths for old and new names
                old_path = os.path.join(class_dir, file)
                new_path = os.path.join(class_dir, new_name)

                # Rename the file
                os.rename(old_path, new_path)

                # Increment the counter
                counter += 1
```

Renaming the images to find them easily.

```python
# Rename images in both training and testing directories
rename_images_in_dir(train_dir)
rename_images_in_dir(test_dir)

print("Image renaming completed.")
```

The image size to match EfficientNetB2 (260x260) and prepares datasets for training and validation. It loads images from the specified directory, shuffling and splitting them into training and validation sets with a batch size of 32. The images are resized to 260x260 before being fed into the model.

```python
# IMG_SIZE to match EfficientNetB2
IMG_SIZE = (260, 260)  # Consistent input size
IMG_SHAPE = IMG_SIZE + (3,)
BATCH_SIZE = 32
data_dir = "/content/drive/MyDrive/AAI/Classification/Data/Training"
# Rebuild datasets with the updated IMG_SIZE

print("train_dataset\n")

train_dataset = image_dataset_from_directory(data_dir,
                                             shuffle=True,
                                             label_mode='categorical',
                                             validation_split=0.2,
                                             batch_size=BATCH_SIZE,
                                             seed=42,
                                             subset="training",
                                             image_size=IMG_SIZE)
print("\nvalidation_dataset\n")
validation_dataset = image_dataset_from_directory(data_dir,
                                                  shuffle=True,
                                                  label_mode='categorical',
                                                  validation_split=0.2,
                                                  batch_size=BATCH_SIZE,
                                                  seed=42,
                                                  subset="validation",
                                                  image_size=IMG_SIZE)

print("\ntest_dataset\n")
test_dataset = image_dataset_from_directory(test_dir,
                                                  shuffle=False,
                                                  label_mode='categorical',
                                                  batch_size=BATCH_SIZE,
                                                  seed=42,
                                                  image_size=IMG_SIZE)
```

Just print the types of the brain tumors in the dataset.

```python
class_names = train_dataset.class_names
num_classes = len(class_names)

for i in range(1, num_classes + 1):
    print(str(i) + ". ", class_names[i - 1])
```

The code divides the validation dataset into two parts: one for validation and a smaller portion for testing. It then prepares the datasets by prefetching them, which helps speed up the training process by loading data in advance. Finally, it prints out how many batches are in the validation and test sets to give a clear idea of the dataset size.

```python
val_batches = tf.data.experimental.cardinality(validation_dataset)
validation_dataset = validation_dataset.skip(val_batches // 5)

print('Number of validation batches: %d' % tf.data.experimental.cardinality(validation_dataset))
print('Number of test batches: %d' % tf.data.experimental.cardinality(test_dataset))

AUTOTUNE = tf.data.AUTOTUNE

train_dataset = train_dataset.prefetch(buffer_size=AUTOTUNE)
validation_dataset = validation_dataset.prefetch(buffer_size=AUTOTUNE)
test_dataset = test_dataset.prefetch(buffer_size=AUTOTUNE)
```

Data augmentation to randomly flip and rotate images, helping the model generalize better. It also updates the prediction layer to match the number of classes, using a softmax activation to output class probabilities.

```python
# Update data_augmentation to match the new IMG_SIZE
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip('horizontal'),
    tf.keras.layers.RandomRotation(0.2),
])

# Update the prediction_layer to match the output shape of EfficientNetB2
num_classes = len(class_names)  # Assuming class_names contains the list of class labels
prediction_layer = tf.keras.layers.Dense(num_classes, activation='softmax')
```

This loads the pre-trained EfficientNetB2 model without the top classification layer, using ImageNet weights for feature extraction. It also sets the input shape to match the expected dimensions (IMG_SHAPE) for the model.

```python
base_model = EfficientNetB2(
    include_top=False,  # Exclude the top classification layer
    weights="imagenet",  # Use pre-trained ImageNet weights
    input_shape=IMG_SHAPE
)
```

This function preprocesses images by applying the preprocess_input method, which prepares the image data for input into the EfficientNetB2 model, ensuring it matches the expected format for the pre-trained model.

This line ensures that the base EfficientNetB2 model's weights are frozen and not updated during training, allowing only the custom layers added on top to be trained initially.

```python
# Ensure the base model is not trainable initially
base_model.trainable = False
```

The code builds a model using EfficientNetB2 by defining an input layer with the specified image shape. It applies data augmentation and preprocessing specific to EfficientNet, followed by passing the image through the pre-trained base model with frozen weights. The output is then processed with global average pooling and dropout for regularization, before passing through the final prediction layer with a softmax activation to classify the image into one of the tumor types.

```python
# Build the model with EfficientNetB2
inputs = tf.keras.Input(shape=IMG_SHAPE)
x = data_augmentation(inputs)         # Apply data augmentation
x = preprocess_input(x)               # Apply EfficientNet preprocessing
x = base_model(x, training=False)     # EfficientNetB2 base model
x = tf.keras.layers.GlobalAveragePooling2D()(x)  # Pooling layer
x = tf.keras.layers.Dropout(0.2)(x)   # Regularization
outputs = prediction_layer(x)         # Final dense layer with correct shape
model = tf.keras.Model(inputs, outputs)
```

- Categorical Accuracy: Measures how often the predicted class matches the true class.
- Precision: Measures the percentage of positive predictions that were actually correct.
- Recall: Measures the percentage of actual positive instances that were correctly identified.

```python
METRICS = [
      metrics.CategoricalAccuracy(name='accuracy'),
      metrics.Precision(name='precision'),
      metrics.Recall(name='recall')
]
```

Below code compiles the model by setting the optimizer to Adam with a learning rate of 0.001, the loss function to Categorical Crossentropy (for multi-class classification), and the evaluation metrics to accuracy, precision, and recall. It prepares the model for training.

```python
# Compile the model
base_learning_rate = 0.001
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=base_learning_rate),
              loss=tf.keras.losses.CategoricalCrossentropy(from_logits=False),
              metrics=METRICS)
```

The model is trained for 10 epochs using the specified training and validation datasets.

```python
# Train the model
initial_epochs = 40
history = model.fit(train_dataset,
                    epochs=initial_epochs,
                    validation_data=validation_dataset)
```

The model is evaluated on the test dataset, and the evaluation results (such as loss and metrics) are printed.

```python
# Evaluate the model
result = model.evaluate(test_dataset)
print(result)
```

The code first prints the loss, accuracy, precision, and recall metrics. Then, it calculates the F1 score using the formula, which combines precision and recall. Finally, it prints all metrics, including the calculated F1 score.

### Predict the Brain Tumor Class
The function predict_image that takes an image path and a trained model as input. It processes the image to match the model’s input requirements, then predicts the class of the image using the model. The predicted class and the model's confidence score are returned and printed.

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.efficientnet import preprocess_input

# Function to predict a single image
def predict_image(image_path, model):
    # Load and preprocess the image
    img = load_img(image_path, target_size=(260, 260))  # Match model input size
    img_array = img_to_array(img)  # Convert image to array
    img_array = preprocess_input(img_array)  # Preprocess to match training conditions
    img_array = tf.expand_dims(img_array, axis=0)  # Add batch dimension

    # Predict the class
    predictions = model.predict(img_array)
    predicted_class = class_names[tf.argmax(predictions[0]).numpy()]
    confidence = tf.reduce_max(predictions[0]).numpy()

    return predicted_class, confidence

# Example usage
image_path = "/content/drive/MyDrive/AAI/Classification/Data/Testing/pituitary_tumor/pituitary_tumor_11.jpg"
predicted_class, confidence = predict_image(image_path, model)
print(f"Predicted class: {predicted_class}, Confidence: {confidence:.2f}")
```

### Plot the training and validation metrics

The plot_training_metrics function plots training and validation metrics (like loss, accuracy, precision, recall) over epochs using matplotlib. It checks if the metrics exist in the training history and visualizes them. If any metric is missing, it skips plotting for that metric.

```python
import matplotlib.pyplot as plt

def plot_training_metrics(history, metrics=None):
    """
    Function to plot training and validation metrics over epochs and add the best epoch point.
    :param history: History object returned from model.fit().
    :param metrics: List of metrics to plot. If None, defaults to common metrics.
    """
    if metrics is None:
        metrics = ['loss', 'accuracy', 'precision', 'recall']

    available_metrics = history.history.keys()  # Keys in the history object

    for metric in metrics:
        # Check if the metric exists in the history
        if metric in available_metrics and f'val_{metric}' in available_metrics:
            plt.figure(figsize=(8, 6))

            # Training and validation metric plots
            plt.plot(history.history[metric], label=f'Training {metric}')
            plt.plot(history.history[f'val_{metric}'], label=f'Validation {metric}')

            # Find the best epoch for validation metric (max for accuracy, min for loss)
            if metric == 'loss':
                best_epoch = min(range(len(history.history[f'val_{metric}'])), key=lambda i: history.history[f'val_{metric}'][i])
                best_value = history.history[f'val_{metric}'][best_epoch]
                plt.scatter(best_epoch, best_value, color='red', zorder=5, label=f'Best Epoch {best_epoch+1} ({best_value:.4f})')
            else:
                best_epoch = max(range(len(history.history[f'val_{metric}'])), key=lambda i: history.history[f'val_{metric}'][i])
                best_value = history.history[f'val_{metric}'][best_epoch]
                plt.scatter(best_epoch, best_value, color='red', zorder=5, label=f'Best Epoch {best_epoch+1} ({best_value:.4f})')

            # Adding labels and title
            plt.xlabel('Epochs')
            plt.ylabel(metric.capitalize())
            plt.title(f'Training and Validation {metric.capitalize()}')
            plt.legend()
            plt.grid(True)
            plt.show()
        else:
            print(f"Metric '{metric}' not found in history. Skipping...")

# `history` contains the model.fit() result
plot_training_metrics(history)
```

This code defines a function create_images_and_masks that takes image and mask file paths as inputs and prepares datasets for training a model. It resizes images and masks to a fixed size (256x256) and stores them in numpy arrays. It handles color images and binary masks, ensuring they are resized and aligned properly. The function returns two arrays: X (images) and Y (masks), with their shapes printed for verification.

```python
IMG_SIZE = 256

def create_images_and_masks(images_path, masks_path, mask_image_type = ".png"):

    if not mask_image_type.startswith("."):
        mask_image_type = "." + mask_image_type

    # extract images and masks ids
    images_ids = next(os.walk(images_path))[2]
    masks_ids = next(os.walk(masks_path))[2]
    # print number of images and masks
    print(f"Total images: {len(images_ids)}\nTotal masks: {len(masks_ids)}")

    # Creating blank array of the length of image_ids i.e 290, 256, 256, 3
    # for color images
    X = np.zeros((len(images_ids), IMG_SIZE, IMG_SIZE, 3), dtype=np.uint8)
    # for binary mask
    Y = np.zeros((len(masks_ids), IMG_SIZE, IMG_SIZE, 1), dtype=np.bool_)

    print("Resizing training images and masks")
    for index, id_ in tqdm(enumerate(images_ids), total=len(images_ids)):
        path = images_path + "/" + id_
        # open and resize the images
        img = imread(path)
        img = resize(img, (IMG_SIZE, IMG_SIZE, 3), mode="constant", preserve_range=True)
        X[index] = img
        # create the mask
        mask = imread(masks_path + "/" + id_.split(".")[0] + mask_image_type)
        mask = np.expand_dims(resize(mask, (IMG_SIZE, IMG_SIZE), mode="constant", preserve_range=True), axis=-1)
        Y[index] = mask

    # Printing the shape of the data
    print(f"\nImages shape: {X.shape}\nMasks shape: {Y.shape}")

    return X, Y
```

```python
# This function take training model and plot history of accuracy and losses with the best epoch in both of them.

def plot_training(hist):

    # Define needed variables
    tr_acc = hist.history['iou_score']
    tr_loss = hist.history['loss']
    val_acc = hist.history['val_iou_score']
    val_loss = hist.history['val_loss']
    index_loss = np.argmin(val_loss)
    val_lowest = val_loss[index_loss]
    index_acc = np.argmax(val_acc)
    acc_highest = val_acc[index_acc]
    Epochs = [i+1 for i in range(len(tr_acc))]
    loss_label = f'best epoch= {str(index_loss + 1)}'
    acc_label = f'best epoch= {str(index_acc + 1)}'

    # Plot training history
    plt.figure(figsize= (20, 8))
    plt.style.use('fivethirtyeight')

    plt.subplot(1, 2, 1)
    plt.plot(Epochs, tr_loss, 'r', label= 'Training loss')
    plt.plot(Epochs, val_loss, 'g', label= 'Validation loss')
    plt.scatter(index_loss + 1, val_lowest, s= 150, c= 'blue', label= loss_label)
    plt.title('Training and Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(Epochs, tr_acc, 'r', label= 'Training IOU Score')
    plt.plot(Epochs, val_acc, 'g', label= 'Validation IOU Score')
    plt.scatter(index_acc + 1 , acc_highest, s= 150, c= 'blue', label= acc_label)
    plt.title('Training and Validation IOU Score')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.tight_layout
    plt.show()


plot_training(results)
```

### **IoU Score for Brain Tumor Segmentation**

The Intersection over Union (IoU) score measures how well the model segments the brain tumor. The model achieved an IoU score of 0.7403603792190552 . This means there is a 73.5% overlap between the predicted tumor and the ground truth.

The IoU score is calculated as:

$$
\text{IoU} = \frac{TP}{TP + FP + FN}
$$

Where:
- \( TP \) = True Positives (correctly predicted tumor pixels)
- \( FP \) = False Positives (background pixels wrongly predicted as tumor)
- \( FN \) = False Negatives (tumor pixels wrongly predicted as background)


- An IoU score of 0.7403603792190552 shows good performance.
- The low loss value (0.022234363481402397) indicates that the predictions are close to actual values.
